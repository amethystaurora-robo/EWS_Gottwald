```python

from matplotlib.lines import Line2D

colors = ['#006d77', '#83c5be', '#ffd166', '#ff7b00', '#e63946', '#457b9d','#6a4c93', '#2a9d8f','#8ac926']
sigmas = [0.9, 0.92, 0.93, 0.94, 0.95, 0.96, 0.97, 0.98, 0.99]

fig, ax = plt.subplots(figsize=(7,5))

# solid lines = on -> off, dashed = off -> on
ax.plot(path_one['S'],   path_one['T'],   color=colors[0], linestyle='-')
ax.plot(path_two['S'],   path_two['T'],   color=colors[1], linestyle='-')
ax.plot(path_three['S'], path_three['T'], color=colors[2], linestyle='-')
ax.plot(path_four['S'],  path_four['T'],  color=colors[3], linestyle='-')
ax.plot(path_five['S'],  path_five['T'],  color=colors[4], linestyle='-')
ax.plot(path_six['S'],  path_six['T'],  color=colors[5], linestyle='-')
ax.plot(path_seven['S'],  path_seven['T'],  color=colors[6], linestyle='-')
ax.plot(path_eight['S'],  path_eight['T'],  color=colors[7], linestyle='-')
ax.plot(path_nine['S'],  path_nine['T'],  color=colors[8], linestyle='-')

#ax.plot(back_one['S'],   back_one['T'],   color=colors[0], linestyle='--')
#ax.plot(back_two['S'],   back_two['T'],   color=colors[1], linestyle='--')
#ax.plot(back_three['S'], back_three['T'], color=colors[2], linestyle='--')
#ax.plot(back_four['S'],  back_four['T'],  color=colors[3], linestyle='--')
#ax.plot(back_five['S'],  back_five['T'],  color=colors[4], linestyle='--')
#ax.plot(back_six['S'],  back_six['T'],  color=colors[5], linestyle='--')
#ax.plot(back_seven['S'],  back_seven['T'],  color=colors[6], linestyle='--')
#ax.plot(back_eight['S'],  back_eight['T'],  color=colors[7], linestyle='--')
#ax.plot(back_nine['S'],  back_nine['T'],  color=colors[8], linestyle='--')

ax.set_xlabel('Salinity (scaled)')
ax.set_ylabel('Temperature (scaled)')
ax.set_title('Instanton Path with Varied Freshwater Flux')

# create legend handles for colors (sigma values)
color_handles = [Line2D([0],[0], color=colors[i], lw=3, label=f'{sigmas[i]}')
                 for i in range(len(sigmas))]

# create legend handles for linestyles (meaning)
style_handles = [
    Line2D([0],[0], color='k', lw=2, linestyle='-', label='on → off'),
    #Line2D([0],[0], color='k', lw=2, linestyle='--', label='off → on')
]
ax.grid(color='gray', linestyle='--', linewidth=0.5, alpha=0.4)
# combine and place legend
first_legend = ax.legend(handles=color_handles, title='sigma', loc='upper left', bbox_to_anchor=(1.02,1))
ax.add_artist(first_legend)
ax.legend(handles=style_handles, title='direction', loc='lower left', bbox_to_anchor=(1.02,0))

plt.tight_layout()
plt.savefig("paths_sigma_towards_stochastic.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
'''
Lineplots of Trajectory shifts
'''

folder_path='../data/sde_paths/'
for file in os.listdir(folder_path):
    if file.endswith(".csv"):
        print(file)
        

        #get variables from file name
        name = file.replace(".csv", "")
        _,  number = name.split("_")

        number  = number.replace("number", "")

        df = pd.read_csv(os.path.join(folder_path, file))

        plt.figure()
        #Lineplot
        sns.set_style("darkgrid")
        sns.lineplot(
            data=df,
            x="time",
            y="T",
            hue="traj",
            estimator=None,          #prevents averaging / collapsing
            lw=1,
            palette=sns.color_palette("Spectral", n_colors=df["traj"].nunique()),
            legend=False)

        plt.title(f"SDE Timeseries Number {number}")
        plt.xlabel("time")
        plt.ylabel("Salinity")
```
